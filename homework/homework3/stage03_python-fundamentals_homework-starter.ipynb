## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit
```

```python
import numpy as np

nums = np.arange(1,1_000_001)

#Loop
def loop_double(a): 
    out = []
    for x in a:
        out.append(x*2)
    return out

#Vector
def vec_double(a):
    return a*2

print("两种写法结果一致吗:", np.array_equal(loop_double(nums), vec_double(nums)))
```

```python
%timeit loop_double(nums)
```

```python
%timeit vec_double(nums)
```

```python
##Example that is more related to quant analyst
import numpy as np

# 假装这是 20 年的每日收盘价(约 5000 个交易日)
rng = np.random.default_rng(42)
prices = 100 * np.cumprod(1 + rng.normal(0.0003, 0.01, 5000))

# 写法 A:循环 —— 一天一天地算
def returns_loop(p):
    out = []
    for i in range(1, len(p)):
        out.append((p[i] - p[i-1]) / p[i-1])
    return np.array(out)

# 写法 B:向量化 —— 整条序列一次算完
def returns_vec(p):
    return p[1:] / p[:-1] - 1

print("两种写法结果一致吗:", np.allclose(returns_loop(prices), returns_vec(prices)))
```

```python
%timeit returns_loop(prices)
```

```python
%timeit returns_vec(prices)
```
